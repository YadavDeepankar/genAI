```python

# Load a pre-trained embedding model.
#
# Model: all-MiniLM-L6-v2
# - Small and fast
# - Produces 384-dimensional embeddings
# - Good for semantic search, clustering, similarity matching
#
# The model is downloaded from Hugging Face the first time.
```

```python
sentences = ["This is an example sentence", "Each sentence is converted"]

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings = model.encode(sentences)
print(embeddings)
```

```python
# Convert sentences into numerical vectors (embeddings).
#
# Output shape:
# (number_of_sentences, embedding_dimension)
#
# For this model:
# (2, 384)
```

```python
senetences = ['I like data science machine learnining',
             'DS ML AI is awesome']

embeddings = model.encode(sentences)
print(embeddings)
```

```python
######## NOTES #########

# Notes

## Objective
Learn how to generate sentence embeddings using Sentence Transformers.

## What is SentenceTransformer?

SentenceTransformer is a library built on top of transformers that converts
text into dense vector representations called embeddings.

These embeddings capture semantic meaning rather than exact words.

Example:

Sentence 1:
"I like machine learning"

Sentence 2:
"I enjoy studying AI"

Even though the words are different, their embeddings will be close because
their meanings are similar.

---

## Model Used

Model:
sentence-transformers/all-MiniLM-L6-v2

Characteristics:
- Embedding dimension: 384
- Lightweight and fast
- Suitable for:
  - Semantic Search
  - Resume Matching
  - RAG Applications
  - Clustering
  - Recommendation Systems

---

## Code Flow

1. Install sentence-transformers library.
2. Import SentenceTransformer class.
3. Load pre-trained embedding model.
4. Provide input sentences.
5. Generate embeddings using model.encode().
6. Print resulting vectors.

---

## Understanding model.encode()

Input:

[
    "This is an example sentence",
    "Each sentence is converted"
]

Output:

[
    [0.12, -0.45, 0.67, ...],
    [0.98, 0.31, -0.72, ...]
]

Each sentence becomes a vector containing 384 numerical values.

Shape:

(number_of_sentences, embedding_dimension)

For this model:

(2, 384)

---

## Important Observation

In the second code block:

senetences = [
    'I like data science machine learnining',
    'DS ML AI is awesome'
]

A new variable is created.

However, the code calls:

model.encode(sentences)

instead of:

model.encode(senetences)

As a result, the first set of sentences is encoded again.

Correct version:

embeddings = model.encode(senetences)

---

## Why Embeddings Matter in GenAI

Embeddings allow comparison based on meaning rather than exact words.

Example:

Query:
"Python developer with data engineering skills"

Resume:
"Built ETL pipelines using Python and Spark"

Keyword matching may miss the connection.

Embedding similarity can identify that both texts discuss related concepts.

This capability powers:
- RAG
- Semantic Search
- Vector Databases
- Resume Classification
- Recommendation Engines

---

## Real-World Use Case: Resume Classification

Approach:

Resume
    ↓
Extract Text
    ↓
Generate Embedding
    ↓
Compare Against Job Role Embeddings
    ↓
Find Top Matching Roles

Example roles:
- Data Engineer
- Data Analyst
- Machine Learning Engineer
- Backend Developer

The role with highest similarity score is considered the best match.

No model training is required.

---

## Key Concepts Learned

1. Embedding
   Numerical representation of text meaning.

2. Embedding Model
   Model that converts text into vectors.

3. Semantic Similarity
   Measure of how close meanings are.

4. Vector Space
   Mathematical space where embeddings are stored.

5. Cosine Similarity
   Most common metric used to compare embeddings.

---

## Interview Questions

Q1. What is an embedding?

A numerical vector representation of data that captures semantic meaning.

Q2. Why use embeddings instead of keyword matching?

Embeddings understand meaning and context, while keyword matching relies on exact words.

Q3. What is the output dimension of all-MiniLM-L6-v2?

384 dimensions.

Q4. What are common applications of embeddings?

- Semantic Search
- RAG
- Recommendation Systems
- Resume Matching
- Clustering

Q5. Does generating embeddings require model training?

No. Pre-trained embedding models can be used directly.
```
